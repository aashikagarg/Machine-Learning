```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("mlbSalaries.csv")
```

```python
# First 5 rows
df.head()

# Dataset shape
df.shape

# Data types and non-null info
df.info()

# Summary statistics
df.describe()

# Check for null values
df.isnull().sum()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['salary'], bins=50, kde=True)
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Assuming your 'df' DataFrame has a 'year' column and a 'salary' column
# Grouping by 'year' and calculating the mean salary for each year
avg_salary_by_year = df.groupby("year")["salary"].mean()

# Plotting the average salary over the years
avg_salary_by_year.plot(kind="line", figsize=(10,5), marker='o')

plt.title("Average Salary Over the Years")
plt.xlabel("Year")
plt.ylabel("Average Salary")
plt.grid(True)
plt.show()
```
