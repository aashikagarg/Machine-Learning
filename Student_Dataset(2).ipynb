```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('student_data.csv')
```

```python
# Display first 5 rows
df.head()

# Display dataset info
df.info()

# Summary statistics
df.describe()

# Check for missing values
df.isnull().sum()
```

```python
# Histogram of final grades (G3)
plt.figure(figsize=(8, 6))
sns.histplot(df['G3'], bins=20, kde=True)
plt.title('Distribution of Final Grades (G3)')
plt.show()

# Boxplot of G3 by sex
sns.boxplot(x='sex', y='G3', data=df)
plt.title('Final Grades by Gender')
plt.show()

# Scatter plot of G1 vs G3
sns.scatterplot(x='G1', y='G3', data=df)
plt.title('First Period Grade vs Final Grade')
plt.show()

```

```python
# Average grades by school
df.groupby('school')['G3'].mean()

# Average grades by parents' education level
df.groupby('Medu')['G3'].mean().plot(kind='bar')
plt.title('Average Final Grade by Mother\'s Education')
plt.ylabel('Average G3 Score')
plt.show()

# Count of students by address type
df['address'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Student Distribution by Address Type')
plt.show()
```

```python
# Regression plot of study time vs final grade
sns.regplot(x='studytime', y='G3', data=df)
plt.title('Study Time vs Final Grade')
plt.show()

# Pairplot of selected numerical features
sns.pairplot(df[['G1', 'G2', 'G3', 'studytime', 'failures', 'absences']])
plt.show()

# Violin plot of grades by internet access
sns.violinplot(x='internet', y='G3', data=df)
plt.title('Final Grades by Internet Access at Home')
plt.show()
```
