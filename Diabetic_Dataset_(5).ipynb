```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.countplot(data=df, x='readmitted', palette='Set2')
plt.title('Patient Readmission Status')
plt.xlabel('Readmitted')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='gender', hue='readmitted', palette='Set1')
plt.title('Readmission by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
df['race'].value_counts().plot.pie(autopct='%1.1f%%', figsize=(6,6), title='Race Distribution')
plt.ylabel('')
plt.show()
```
